```python
import torch
import mlflow

from mlflow import MlflowClient

# Auto log all MLflow entities
mlflow.pytorch.autolog(silent=True)
mlflow.set_tracking_uri("http://127.0.0.1:5000")
```

```python
# get the experiment ID if it exists, otherwise create it
def get_or_create_experiment(experiment_name):
    if experiment := mlflow.get_experiment_by_name(experiment_name):
        return experiment.experiment_id
    else:
        return mlflow.create_experiment(experiment_name)
    
```

```python
from mlflow import log_metric, log_param, log_artifacts

# Create a new MLflow Experiment
experiment_name = "Test"
run_name = "test"
experiment_id = get_or_create_experiment(experiment_name)

mlflow.set_experiment(experiment_id=experiment_id)

# main run
with mlflow.start_run(experiment_id=experiment_id, run_name='test') as parent_run:
    
    for i in range(10):
        with mlflow.start_run(nested=True):
            cfg = dict(
                alpha=0.1,
                beta=0.2,
                gamma=0.3
            )
            # test logging
            mlflow.log_params(cfg)
        print(f"Ending nested run {i}")
    print("Ending parent run")
```
